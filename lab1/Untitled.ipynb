```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def show_image(image, title="", cmap=None):
    plt.figure(figsize=(6, 6))
    plt.title(title)

    if cmap is None:
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    plt.imshow(image, cmap=cmap)
    plt.axis('off') 
    plt.show()
```

### Задание 1. Загрузить цветное изображение

```python
img = cv2.imread('DQxwyWmAI1.png')

if img is None:
    print("Ошибка: Изображение не найдено! Проверь имя файла и путь.")
else:
    show_image(img, "1. Исходное цветное изображение")
```

### Задание 2. Преобразовать изображение в градации серого и вывести минимальное и максимальное значение пикселей.

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

min_val = img_gray.min()
max_val = img_gray.max()

print(f"Минимальное значение пикселя: {min_val}")
print(f"Максимальное значение пикселя: {max_val}")

show_image(img_gray, "2. Градации серого", cmap='gray')
```

---
### Задание 3. Преобразовать изображение в YCrCb и вывести отдельные каналы.

```python
img_ycrcb = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)

y, cr, cb = cv2.split(img_ycrcb)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(y, cmap='gray')
axes[0].set_title("Канал Y (Яркость)")
axes[0].axis('off')

axes[1].imshow(cr, cmap='gray')
axes[1].set_title("Канал Cr (Разность красного)")
axes[1].axis('off')

axes[2].imshow(cb, cmap='gray')
axes[2].set_title("Канал Cb (Разность синего)")
axes[2].axis('off')

plt.show()
```

---
### Задание 4. Сделать инверсию цвета (отнять каждый канал от 255) и сохранить.

```python
img_inverted = 255 - img

output_filename = 'inverted_image.jpg'
cv2.imwrite(output_filename, img_inverted)

print(f"Инвертированное изображение сохранено в папку с блокнотом как '{output_filename}'")
show_image(img_inverted, "4. Инверсия цвета")
```

### Задание 5. Нарисовать прямоугольник на изображении красного цвета вокруг центральной области.

```python
img_rect = img.copy()

height, width = img_rect.shape[:2]

center_x, center_y = width // 2, height // 2

offset_x = 100 
offset_y = 50  

pt1 = (center_x - offset_x, center_y - offset_y)
pt2 = (center_x + offset_x, center_y + offset_y)

color = (0, 0, 255)
thickness = 3

cv2.rectangle(img_rect, pt1, pt2, color, thickness)

show_image(img_rect, "5. Красный прямоугольник в центре")
```
